```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_excel('/content/cleaned_dataset_for_data_analytics.xlsx')
```

```python
df['Date'] = pd.to_datetime(df['Date'])
```

```python
monthly_sales = df.groupby(df['Date'].dt.month)['TotalPrice'].sum()

monthly_sales.plot(kind='line', marker='o', figsize=(8,5))

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.grid(True)

plt.show()
```

```python
top_products = df.groupby('Product')['Quantity'].sum().sort_values(ascending=False)

top_products.plot(kind='bar', figsize=(10,5))

plt.title("Top Selling Products")
plt.xlabel("Product")
plt.ylabel("Quantity Sold")

plt.show()
```

```python
df['PaymentMethod'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    figsize=(6,6)
)

plt.title("Payment Method Distribution")

plt.ylabel('')

plt.show()
```

```python
sns.countplot(x='OrderStatus', data=df)

plt.title("Order Status Distribution")

plt.xticks(rotation=45)

plt.show()
```

```python
numeric_df = df.select_dtypes(include='number')

plt.figure(figsize=(8,5))

sns.heatmap(numeric_df.corr(), annot=True)

plt.title("Correlation Heatmap")

plt.show()
```

```python
category_sales = df.groupby('Product')['TotalPrice'].sum()

category_sales.plot(kind='bar')

plt.title("Revenue by Product")
plt.xlabel("Product")
plt.ylabel("Revenue")

plt.show()
```

```python
print("Total Revenue :", df['TotalPrice'].sum())

print("Average Order Value :", df['TotalPrice'].mean())

print("Highest Order :", df['TotalPrice'].max())

print("Total Orders :", df['OrderID'].count())
```
