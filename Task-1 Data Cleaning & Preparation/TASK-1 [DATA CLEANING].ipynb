```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel("Dataset for Data Analytics.xlsx")
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df["CouponCode"] = df["CouponCode"].fillna("No Coupon")
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df["OrderID"].duplicated().sum()
```

```python
df["Date"] = pd.to_datetime(df["Date"])
```

```python
df["Date"] = df["Date"].dt.strftime("%Y-%m-%d")
```

```python
df["Product"] = df["Product"].str.title().str.strip()

df["PaymentMethod"] = df["PaymentMethod"].str.title().str.strip()

df["ReferralSource"] = df["ReferralSource"].str.title().str.strip()

df["OrderStatus"] = df["OrderStatus"].str.title().str.strip()
```

```python
df["UnitPrice"] = df["UnitPrice"].round(2)

df["TotalPrice"] = df["TotalPrice"].round(2)

df["CalculatedTotal"] = (df["Quantity"] * df["UnitPrice"]).round(2)
```

```python
df[["Quantity", "UnitPrice", "TotalPrice", "CalculatedTotal"]].head()
```

```python
incorrect_totals = df[df["TotalPrice"] != df["CalculatedTotal"]]

incorrect_totals.head()
```

```python
df.loc[df["TotalPrice"] != df["CalculatedTotal"], "TotalPrice"] = df["CalculatedTotal"]
```

```python
incorrect_totals = df[df["TotalPrice"] != df["CalculatedTotal"]]

incorrect_totals.head()
```

```python
df["TotalPrice"] = df["CalculatedTotal"]
```

```python
print(df.isnull().sum())

print(df.duplicated().sum())

print(df["OrderID"].duplicated().sum())
```

```python
df.to_excel("cleaned_dataset_for_data_analytics.xlsx", index=False)
```
